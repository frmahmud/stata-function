# Stata-like Analysis in Python
This notebook demonstrates how to use the custom `codebook` and `tab` functions on Stata's built-in `auto.dta` dataset.

```python
import pandas as pd
import numpy as np
from statalikefunction import read_dta, codebook, tab

# Load the dataset
df = read_dta("auto.dta")

# Display dataset information
print(f"Dataset: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"\nVariables: {df.columns.tolist()}")
```

## Codebook for Python
Replicates Stata's `codebook` command — shows variable label, type, missing values. 

```python
# Categorical variable (value-labelled in Stata)
codebook(df, 'make')
```

```python
# Numeric variable — price
codebook(df, 'price')
```

```python
# Categorical variable — foreign (domestic vs foreign)
codebook(df, 'foreign')
```

```python
# Numeric variable — mpg (miles per gallon)
codebook(df, 'mpg')
```

## Tabulation for Python
Replicates Stata's `tab` command — shows frequency, percent, and cumulative percent.

```python
# Tab on rep78 (repair record — has missing values, good test case)
tab(df, 'mpg')
```

```python
# Tab on foreign (domestic vs foreign)
tab(df, 'foreign')
```
